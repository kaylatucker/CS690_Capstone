```python
from pathlib import Path
import pandas as pd

# If the notebook is inside 03_notebooks/
PROJECT_ROOT = Path("..")
RAW_DIR = PROJECT_ROOT / "02_data" / "raw"

csv_files = sorted(RAW_DIR.glob("*.csv"))

print(f"Found {len(csv_files)} CSV files:")
for file in csv_files:
    print("-", file.name)
```

```python
for file in csv_files:
    df = pd.read_csv(file)

    print("\n" + "=" * 80)
    print(file.name)
    print("=" * 80)

    print(f"Rows: {df.shape[0]:,}")
    print(f"Columns: {df.shape[1]}")
    
    print("\nColumn names:")
    print(df.columns.tolist())

    print("\nData types:")
    print(df.dtypes)

    print("\nMissing values:")
    missing = pd.DataFrame({
        "missing_count": df.isna().sum(),
        "missing_percent": (df.isna().mean() * 100).round(2)
    })
    print(missing[missing["missing_count"] > 0])

    print(f"\nExact duplicate rows: {df.duplicated().sum():,}")

    print("\nFirst 3 rows:")
    display(df.head(3))

    del df
```

```python
key_checks = {
    "olist_customers_dataset.csv": ["customer_id"],
    "olist_orders_dataset.csv": ["order_id"],
    "olist_order_items_dataset.csv": ["order_id", "order_item_id"],
    "olist_products_dataset.csv": ["product_id"],
    "olist_sellers_dataset.csv": ["seller_id"],
    "olist_order_payments_dataset.csv": ["order_id", "payment_sequential"],
    "olist_order_reviews_dataset.csv": ["review_id"],
    "product_category_name_translation.csv": ["product_category_name"],
}

for filename, keys in key_checks.items():
    file = RAW_DIR / filename

    if not file.exists():
        print(f"{filename}: FILE NOT FOUND")
        continue

    df = pd.read_csv(file)

    duplicates = df.duplicated(subset=keys).sum()

    print(
        f"{filename}: "
        f"{keys} -> {duplicates:,} duplicate key combination(s)"
    )

    del df
```

```python
for file in csv_files:
    df = pd.read_csv(file, nrows=5)

    date_columns = [
        col for col in df.columns
        if "date" in col.lower() or "timestamp" in col.lower()
    ]

    if date_columns:
        print(f"{file.name}:")
        for col in date_columns:
            print("  -", col)
```

```python
reviews = pd.read_csv(RAW_DIR / "olist_order_reviews_dataset.csv")

duplicate_review_ids = reviews[
    reviews.duplicated(subset=["review_id"], keep=False)
].sort_values("review_id")

print("Rows involved in duplicated review IDs:", len(duplicate_review_ids))
print("Unique duplicated review IDs:", duplicate_review_ids["review_id"].nunique())

display(duplicate_review_ids.head(20))

review_id_order_counts = (
    reviews.groupby("review_id")["order_id"]
    .nunique()
    .sort_values(ascending=False)
)

print(review_id_order_counts.head(10))
```
